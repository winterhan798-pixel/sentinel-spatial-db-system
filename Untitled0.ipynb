```python
import os
import subprocess
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import matplotlib.font_manager as fm
from scipy.interpolate import interp1d
from scipy.signal import detrend
from scipy.stats import chi2

# 1. 自动配置中文字体（防方块乱码）
def setup_chinese_font():
    font_paths = fm.findSystemFonts()
    cjk_fonts = [p for p in font_paths if 'NotoSansCJK' in p or 'WenQuanYi' in p or 'SimHei' in p]
    if not cjk_fonts:
        try:
            subprocess.run(['apt-get', 'update', '-qq'], check=True)
            subprocess.run(['apt-get', 'install', '-y', '-qq', 'fonts-noto-cjk'], check=True)
            fm._load_fontmanager(try_read_cache=False)
            font_paths = fm.findSystemFonts()
            cjk_fonts = [p for p in font_paths if 'NotoSansCJK' in p]
        except Exception:
            pass
    if cjk_fonts:
        return fm.FontProperties(fname=cjk_fonts[0])
    return fm.FontProperties()

zh_font = setup_chinese_font()
plt.rcParams['axes.unicode_minus'] = False

# 2. 读取真实数据（兼容根目录 /、当前目录、/content/）
candidates_gw = ['real_gw_ts.csv', '/real_gw_ts.csv', '/content/real_gw_ts.csv']
candidates_insar = ['real_insar_ts.csv', '/real_insar_ts.csv', '/content/real_insar_ts.csv']

gw_path = next((p for p in candidates_gw if os.path.exists(p)), None)
insar_path = next((p for p in candidates_insar if os.path.exists(p)), None)

if gw_path is None:
    raise FileNotFoundError("没找到 real_gw_ts.csv，请确认文件在 / 或 /content 下")
if insar_path is None:
    raise FileNotFoundError("没找到 real_insar_ts.csv，请确认文件在 / 或 /content 下")

print("读取 real_gw_ts.csv :", gw_path)
print("读取 real_insar_ts.csv:", insar_path, "\n")

df_gw = pd.read_csv(gw_path)
df_insar = pd.read_csv(insar_path)

df_gw['Date'] = pd.to_datetime(df_gw['Date'])
df_insar['Date'] = pd.to_datetime(df_insar['Date'])

# 3. 12天统一采样间隔对齐与季节性振幅质控筛选 (A1 >= 10 mm)
dt = 12.0 / 365.25  # 12天采样步长（单位：年）
common_wells = sorted(list(set(df_gw['Well_ID']).intersection(set(df_insar['Well_ID']))))

valid_wells = {}
for wid in common_wells:
    sub_g = df_gw[df_gw['Well_ID'] == wid].sort_values('Date').drop_duplicates('Date')
    sub_i = df_insar[df_insar['Well_ID'] == wid].sort_values('Date').drop_duplicates('Date')

    start_date = max(sub_g['Date'].min(), sub_i['Date'].min())
    end_date = min(sub_g['Date'].max(), sub_i['Date'].max())
    if (end_date - start_date).days < 730:
        continue

    # 构建 12 天均匀网格，绝不向外外推
    grid_dates = pd.date_range(start=start_date, end=end_date, freq='12D')
    t_grid = (grid_dates - grid_dates[0]).days.values / 365.25

    t_g = (sub_g['Date'] - grid_dates[0]).dt.days.values / 365.25
    t_i = (sub_i['Date'] - grid_dates[0]).dt.days.values / 365.25


    f_g = interp1d(t_g, sub_g['GW_level_m'].values, kind='linear')
    f_i = interp1d(t_i, sub_i['InSAR_mm'].values, kind='linear')

    gw_interp = f_g(t_grid)
    insar_interp = f_i(t_grid)

    if np.std(gw_interp) < 1e-6 or np.std(insar_interp) < 1e-6:
        continue

    # 最小二乘拟合年振幅 A1
    G = np.column_stack([
        np.ones_like(t_grid), t_grid,
        np.cos(2 * np.pi * t_grid), np.sin(2 * np.pi * t_grid),
        np.cos(4 * np.pi * t_grid), np.sin(4 * np.pi * t_grid)
    ])
    beta, _, _, _ = np.linalg.lstsq(G, insar_interp, rcond=None)
    v_fit = beta[1]
    A1_fit = np.hypot(beta[2], beta[3])

    if A1_fit >= 10.0:  # 论文 Section 4.4 质控门槛
        gw_det = detrend(gw_interp)
        insar_det = detrend(insar_interp)
        corr = np.corrcoef(gw_det, insar_det)[0, 1]
        valid_wells[wid] = {
            'Well_ID': wid,
            'Well_Name': sub_i['Well_Name'].iloc[0],
            'Aquifer_Type': sub_i['Aquifer_Type'].iloc[0],
            'dates': grid_dates,
            't': t_grid,
            'gw_raw': gw_interp,
            'insar_raw': insar_interp,
            'gw_det': gw_det,
            'insar_det': insar_det,
            'v_fit': v_fit,
            'A1_fit': A1_fit,
            'corr': corr
        }

print(f"通过 A1 >= 10 mm 质控的高信噪比有效井数量: {len(valid_wells)} 口")
summary_df = pd.DataFrame([{k: v[k] for k in ['Well_ID', 'Well_Name', 'Aquifer_Type', 'v_fit', 'A1_fit', 'corr']} for v in valid_wells.values()])
print(summary_df.sort_values('A1_fit', ascending=False).to_string(index=False))
```

```python
# 标准 Morlet (omega0=6) 连续小波变换 (CWT) 与交叉小波变换 (XWT) 测试验证
def morlet_cwt(signal, dt, dj=0.05, s0=None, J=None, omega0=6.0):
    x = (signal - np.mean(signal)) / np.std(signal)
    N = len(x)
    # 估算 AR(1) 滞后1自相关系数，用于红噪声背景检验
    alpha = np.corrcoef(x[:-1], x[1:])[0, 1]
    alpha = np.clip(alpha, 0.0, 0.99)

    # 零填充到 2 的幂次以消除循环卷积边界伪影
    n_pad = int(2 ** np.ceil(np.log2(N)))
    x_pad = np.pad(x, (0, n_pad - N), mode='constant')

    # 频率轴 (rad / yr)
    freqs = np.fft.fftfreq(n_pad, d=dt) * 2 * np.pi
    x_fft = np.fft.fft(x_pad)

    # Morlet 傅里叶波长因子
    fourier_factor = (4 * np.pi) / (omega0 + np.sqrt(2 + omega0**2))
    if s0 is None:
        s0 = (0.10) / fourier_factor  # 最小周期 ~0.10 年
    if J is None:
        J = int(np.floor(np.log2(2.0 / (s0 * fourier_factor)) / dj))

    scales = s0 * (2 ** (np.arange(0, J + 1) * dj))
    periods = scales * fourier_factor

    W = np.zeros((len(scales), N), dtype=np.complex128)
    for i, s in enumerate(scales):
        # 仅在正频率处构造归一化 Morlet 子波
        daughter = np.zeros(n_pad, dtype=np.complex128)
        pos = freqs > 0
        norm = np.sqrt(2 * np.pi * s / dt) * (np.pi ** -0.25)
        daughter[pos] = norm * np.exp(-0.5 * ((s * freqs[pos] - omega0) ** 2))
        w_full = np.fft.ifft(x_fft * daughter)
        W[i, :] = w_full[:N]

    # 影响锥 (COI) 周期边界 (Torrence & Compo, 1998)
    t_edge = np.minimum(np.arange(N) + 0.5, N - 1 - np.arange(N) + 0.5) * dt
    coi_period = fourier_factor * np.sqrt(2) * t_edge

    # AR(1) 红噪声理论功率谱 P_k
    freq_cycles = 1.0 / periods
    P_k = (1 - alpha**2) / (1 + alpha**2 - 2 * alpha * np.cos(2 * np.pi * freq_cycles * dt))
    # 95% 显著性阈值 (自由度 nu=2)
    signif_95 = P_k * chi2.ppf(0.95, df=2) / 2.0

    return W, periods, coi_period, signif_95, P_k

def cross_wavelet(sig_gw, sig_insar, dt):
    W_gw, periods, coi_gw, _, Pk_gw = morlet_cwt(sig_gw, dt)
    W_in, _, coi_in, _, Pk_in = morlet_cwt(sig_insar, dt)
    coi = np.minimum(coi_gw, coi_in)

    # XWT: W_gw * conj(W_in) -> 相位角 = phase(gw) - phase(insar)
    # 正相位角表示 GW 超前于 InSAR（即 InSAR 滞后于 GW）
    W_xy = W_gw * np.conj(W_in)

    # XWT 95% 显著性检验阈值 (Grinsted et al., 2004, Z_v(95%) = 3.999 for v=2)
    Z_95 = 3.999
    xwt_signif = Z_95 * np.sqrt(Pk_gw * Pk_in)

    # 提取 1 年主周期处的圆平均滞后天数
    idx_1yr = np.argmin(np.abs(periods - 1.0))
    valid_mask = coi >= periods[idx_1yr]
    phases_1yr = np.angle(W_xy[idx_1yr, valid_mask])
    mean_phase = np.angle(np.mean(np.exp(1j * phases_1yr)))
    # 转换为正滞后天数 [0, 365.25)
    lag_days = (mean_phase / (2 * np.pi)) * 365.25
    if lag_days < -90:
        lag_days += 365.25

    return W_gw, W_in, W_xy, periods, coi, xwt_signif, lag_days

# 验证所有有效井的滞后天数
for wid, item in sorted(valid_wells.items(), key=lambda x: -x[1]['A1_fit']):
    _, _, _, _, _, _, lag = cross_wavelet(item['gw_det'], item['insar_det'], dt)
    print(f"{wid} {item['Well_Name']:10s} ({item['Aquifer_Type']:3s}) | A1={item['A1_fit']:5.1f} mm | 滞后={lag:5.1f} 天")
```

```python

# 检查 dt 轴顺序与绘图效果
import matplotlib.gridspec as gridspec

# 选择 3 口极具代表性的高信噪比强耦合井
target_wids = [565306, 565315, 565107]  # 呼芳路17公里(潜水)、祁家湖村黑家塘片区(潜水)、佃坝村五区(承压水)

fig = plt.figure(figsize=(16, 10.5), dpi=180)
outer_gs = gridspec.GridSpec(3, 3, width_ratios=[1.15, 1.05, 1.05], wspace=0.26, hspace=0.32)

levels = np.linspace(-5, 5, 65)
tick_locs = [-5, -3, -1, 0, 1, 3, 5]
tick_labels = ['1/32', '1/8', '1/2', '1', '2', '8', '32']
y_ticks = [0.125, 0.25, 0.5, 1.0]
y_tick_labels = ['0.125', '0.25', '0.5', '1']

for row_idx, wid in enumerate(target_wids):
    item = valid_wells[wid]
    dates = item['dates']
    t_num = mdates.date2num(dates)

    W_gw, periods, coi_gw, sig_gw, Pk_gw = morlet_cwt(item['gw_det'], dt)
    W_in, _, coi_in, sig_in, Pk_in = morlet_cwt(item['insar_det'], dt)
    _, _, W_xy, _, coi_xy, sig_xy, lag_days = cross_wavelet(item['gw_det'], item['insar_det'], dt)

    # --- 左栏：去趋势时序双纵轴对比 ---
    ax_ts = fig.add_subplot(outer_gs[row_idx, 0])
    l1 = ax_ts.plot(dates, item['insar_det'], color='#1f4e78', lw=1.3, marker='o', markersize=2.2, label='地表垂直形变 (去趋势, mm)')
    ax_ts.set_ylabel('地表垂直形变 (mm)', color='#1f4e78', fontproperties=zh_font, fontsize=9.5)
    ax_ts.tick_params(axis='y', labelcolor='#1f4e78', labelsize=8.5)
    ax_ts.grid(True, linestyle='--', alpha=0.35)

    ax_ts2 = ax_ts.twinx()
    l2 = ax_ts2.plot(dates, item['gw_det'], color='#c00000', lw=1.3, linestyle='--', marker='s', markersize=2.0, alpha=0.85, label='地下水位高程 (去趋势, m)')
    ax_ts2.set_ylabel('地下水位高程 (m)', color='#c00000', fontproperties=zh_font, fontsize=9.5)
    ax_ts2.tick_params(axis='y', labelcolor='#c00000', labelsize=8.5)

    ax_ts.set_title(f"({chr(97+row_idx)}) {item['Well_Name']} ({wid} · {item['Aquifer_Type']})", fontproperties=zh_font, fontsize=10.5, fontweight='bold', loc='left', pad=6)
    ax_ts.xaxis.set_major_locator(mdates.YearLocator())
    ax_ts.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
    ax_ts.tick_params(axis='x', labelsize=8.5)
    if row_idx == 0:
        lines = l1 + l2
        labels = [l.get_label() for l in lines]
        ax_ts.legend(lines, labels, loc='upper right', prop=zh_font, fontsize=7.8, framealpha=0.85)

    # --- 中栏：上下子图 (形变 CWT + 水位 CWT) ---
    inner_gs = gridspec.GridSpecFromSubplotSpec(2, 1, subplot_spec=outer_gs[row_idx, 1], hspace=0.18)
    for sub_idx, (W_curr, sig_curr, title_str) in enumerate([
        (W_in, sig_in, '地表形变连续小波谱 (CWT)'),
        (W_gw, sig_gw, '地下水位连续小波谱 (CWT)')
    ]):
        ax_cwt = fig.add_subplot(inner_gs[sub_idx])
        power = np.abs(W_curr) ** 2
        log2_power = np.clip(np.log2(power + 1e-12), -5, 5)

        cf = ax_cwt.contourf(t_num, np.log2(periods), log2_power, levels=levels, cmap='jet', extend='both')
        # 95% 置信度红噪声显著性等值线
        sig_ratio = power / sig_curr[:, None]
        ax_cwt.contour(t_num, np.log2(periods), sig_ratio, levels=[1.0], colors='k', linewidths=1.3)

        # 影响锥 COI 白色半透明遮罩
        ax_cwt.plot(t_num, np.log2(coi_xy), 'k--', lw=0.9)
        ax_cwt.fill_between(t_num, np.log2(coi_xy), np.log2(2.0), color='white', alpha=0.58)

        ax_cwt.set_ylim(np.log2(1.6), np.log2(0.12))
        ax_cwt.set_yticks(np.log2(y_ticks))
        ax_cwt.set_yticklabels(y_tick_labels, fontsize=8)
        ax_cwt.set_ylabel('周期 (年)', fontproperties=zh_font, fontsize=8.5)
        ax_cwt.text(0.03, 0.84, title_str, transform=ax_cwt.transAxes, fontproperties=zh_font, fontsize=8.5,
                    color='white', fontweight='bold', bbox=dict(boxstyle='round,pad=0.2', facecolor='black', alpha=0.55, edgecolor='none'))
        ax_cwt.xaxis.set_major_locator(mdates.YearLocator())
        ax_cwt.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
        if sub_idx == 0:
            ax_cwt.set_xticklabels([])
        else:
            ax_cwt.tick_params(axis='x', labelsize=8.5)
        cbar = plt.colorbar(cf, ax=ax_cwt, ticks=tick_locs, pad=0.02, aspect=14)
        cbar.ax.set_yticklabels(tick_labels, fontsize=7)

    # --- 右栏：交叉小波互功率谱 (XWT) ---
    ax_xwt = fig.add_subplot(outer_gs[row_idx, 2])
    xwt_power = np.abs(W_xy)
    log2_xwt = np.clip(np.log2(xwt_power + 1e-12), -5, 5)

    cf_xwt = ax_xwt.contourf(t_num, np.log2(periods), log2_xwt, levels=levels, cmap='jet', extend='both')
    xwt_sig_ratio = xwt_power / sig_xy[:, None]
    ax_xwt.contour(t_num, np.log2(periods), xwt_sig_ratio, levels=[1.0], colors='k', linewidths=1.5)

    # 绘制相位箭头（向右=同相位，右偏下=形变滞后于水位）
    phase = np.angle(W_xy)
    t_step = max(1, len(t_num) // 22)
    p_step = max(1, len(periods) // 14)
    T_mesh, P_mesh = np.meshgrid(t_num[::t_step], np.log2(periods[::p_step]))
    # 注意：纵轴向下增大，若要表达"正相位滞后(GW超前InSAR)"向下偏转，v 取 -sin(phase)
    U = np.cos(phase[::p_step, ::t_step])
    V = -np.sin(phase[::p_step, ::t_step])
    ax_xwt.quiver(T_mesh, P_mesh, U, V, units='width', scale=28, width=0.0032, headwidth=3.8, headlength=4.5, color='k', alpha=0.85)

    ax_xwt.plot(t_num, np.log2(coi_xy), 'k--', lw=1.0)
    ax_xwt.fill_between(t_num, np.log2(coi_xy), np.log2(2.0), color='white', alpha=0.58)

    ax_xwt.set_ylim(np.log2(1.6), np.log2(0.12))
    ax_xwt.set_yticks(np.log2(y_ticks))
    ax_xwt.set_yticklabels(y_tick_labels, fontsize=8.5)
    ax_xwt.set_ylabel('周期 (年)', fontproperties=zh_font, fontsize=9)
    ax_xwt.set_title('交叉小波互功率谱 (XWT)', fontproperties=zh_font, fontsize=10, fontweight='bold', pad=6)
    ax_xwt.text(0.04, 0.88, f"年周期滞后: {lag_days:.1f} 天", transform=ax_xwt.transAxes, fontproperties=zh_font,
                fontsize=9, color='white', fontweight='bold',
                bbox=dict(boxstyle='round,pad=0.28', facecolor='black', alpha=0.68, edgecolor='none'))
    ax_xwt.xaxis.set_major_locator(mdates.YearLocator())
    ax_xwt.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
    ax_xwt.tick_params(axis='x', labelsize=8.5)

    cbar_xwt = plt.colorbar(cf_xwt, ax=ax_xwt, ticks=tick_locs, pad=0.02, aspect=26)
    cbar_xwt.ax.set_yticklabels(tick_labels, fontsize=7.5)

plt.savefig('wavelet_diagnostic_panel.png', dpi=200, bbox_inches='tight')
print("Saved wavelet_diagnostic_panel.png successfully.")
```
